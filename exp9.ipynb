```python
# ============================================
# Brute-Force Login Attempt Detection
# Google Colab Ready
# ============================================

import re
from collections import Counter
from google.colab import files

# Step 1: Upload the login log file
print("Upload your login log file:")
uploaded = files.upload()

log_file = list(uploaded.keys())[0]

# Step 2: Read the log file
with open(log_file, "r", errors="ignore") as file:
    log_data = file.readlines()

# Step 3: Detect failed login attempts
failed_attempts = []

# Supports common Linux SSH log format:
# Failed password for user from IP_ADDRESS
pattern = r"Failed password for .* from ([0-9]+\.[0-9]+\.[0-9]+\.[0-9]+)"

for line in log_data:
    match = re.search(pattern, line)

    if match:
        ip_address = match.group(1)
        failed_attempts.append(ip_address)

# Step 4: Count failed attempts from each IP address
ip_counts = Counter(failed_attempts)

# Threshold for possible brute-force attack
THRESHOLD = 5

print("\n====================================")
print("BRUTE-FORCE LOGIN ANALYSIS")
print("====================================")

if len(ip_counts) == 0:
    print("No failed login attempts detected.")

else:
    print("\nFailed Login Attempts by IP Address:\n")

    for ip, count in ip_counts.items():
        print(f"IP Address: {ip} | Failed Attempts: {count}")

    print("\n====================================")
    print("POSSIBLE BRUTE-FORCE ATTACKS")
    print("====================================")

    attack_detected = False

    for ip, count in ip_counts.items():

        if count >= THRESHOLD:
            print(
                f"[ALERT] Possible brute-force attack detected "
                f"from {ip} ({count} failed attempts)"
            )
            attack_detected = True

    if not attack_detected:
        print("No brute-force attack detected.")
```
