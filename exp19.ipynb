```python
import hashlib

# Function to calculate SHA-256 hash
def sha256_hash(text):
    return hashlib.sha256(text.encode()).hexdigest()

# Original and modified messages
original_text = "Digital Forensics"
modified_text = "Digital ForensicS"   # Only one character changed

# Calculate hashes
original_hash = sha256_hash(original_text)
modified_hash = sha256_hash(modified_text)

# Display results
print("=" * 60)
print("      HASHING & AVALANCHE EFFECT DEMONSTRATION")
print("=" * 60)

print("\nOriginal Text : ", original_text)
print("SHA-256 Hash  : ", original_hash)

print("\nModified Text : ", modified_text)
print("SHA-256 Hash  : ", modified_hash)

# Compare hashes
print("\nVerification")
print("-" * 60)

if original_hash == modified_hash:
    print("Hashes are identical.")
else:
    print("Hashes are completely different.")
    print("Avalanche Effect Demonstrated Successfully!")

print("=" * 60)
```
