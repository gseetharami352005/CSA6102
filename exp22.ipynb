```python
import math

# Sample file size (in bytes)
file_size = 10000

# Cluster size (in bytes)
cluster_size = 4096

# Calculate number of clusters required
clusters = math.ceil(file_size / cluster_size)

# Calculate allocated space
allocated_space = clusters * cluster_size

# Calculate slack space
slack_space = allocated_space - file_size

print("=" * 50)
print("      SLACK SPACE CALCULATION")
print("=" * 50)
print("File Size          :", file_size, "bytes")
print("Cluster Size       :", cluster_size, "bytes")
print("Clusters Required  :", clusters)
print("Allocated Space    :", allocated_space, "bytes")
print("Slack Space        :", slack_space, "bytes")
print("=" * 50)
```
