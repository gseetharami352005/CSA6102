```python
# File Carving (JPEG Recovery)

# JPEG file signatures
JPEG_HEADER = b'\xFF\xD8'
JPEG_FOOTER = b'\xFF\xD9'

# -----------------------------
# Create a sample binary file
# -----------------------------
with open("disk_image.bin", "wb") as f:
    f.write(b"RandomData12345")
    f.write(JPEG_HEADER)
    f.write(b"This is sample JPEG data")
    f.write(JPEG_FOOTER)
    f.write(b"MoreRandomData67890")

print("Sample disk image created.")

# -----------------------------
# Read the binary file
# -----------------------------
with open("disk_image.bin", "rb") as f:
    data = f.read()

# -----------------------------
# Search for JPEG signatures
# -----------------------------
start = data.find(JPEG_HEADER)
end = data.find(JPEG_FOOTER)

if start != -1 and end != -1:
    end += len(JPEG_FOOTER)

    # Extract carved file
    carved_data = data[start:end]

    with open("recovered_image.jpg", "wb") as out:
        out.write(carved_data)

    print("\nJPEG File Successfully Recovered!")
    print("Recovered File : recovered_image.jpg")
    print("Recovered Size :", len(carved_data), "bytes")

else:
    print("No JPEG file found.")
```
