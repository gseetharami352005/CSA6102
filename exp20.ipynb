```python
# File System Feature Simulation (FAT vs NTFS vs EXT4)

file_systems = {
    "FAT32": {
        "Max File Size": "4 GB",
        "Permissions": "No",
        "Journaling": "No",
        "Encryption": "No",
        "Compression": "No"
    },

    "NTFS": {
        "Max File Size": "16 TB",
        "Permissions": "Yes",
        "Journaling": "Yes",
        "Encryption": "Yes",
        "Compression": "Yes"
    },

    "EXT4": {
        "Max File Size": "16 TB",
        "Permissions": "Yes",
        "Journaling": "Yes",
        "Encryption": "No",
        "Compression": "No"
    }
}

print("=" * 65)
print("       FILE SYSTEM FEATURE SIMULATION")
print("=" * 65)

for fs, features in file_systems.items():
    print("\nFile System:", fs)
    print("-" * 40)

    for feature, value in features.items():
        print(f"{feature:<20}: {value}")

print("\n" + "=" * 65)
print("Simulation Completed Successfully")
print("=" * 65)
```
