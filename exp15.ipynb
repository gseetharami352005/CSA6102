```python
# ============================================
# VOLATILE VS NON-VOLATILE EVIDENCE CAPTURE
# Google Colab Ready
# ============================================

import os
import platform
import hashlib
import socket
import subprocess
from datetime import datetime
from google.colab import files

# --------------------------------------------
# Create Evidence Folder
# --------------------------------------------

evidence_folder = "forensic_evidence"
os.makedirs(evidence_folder, exist_ok=True)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

# ============================================
# 1. VOLATILE EVIDENCE CAPTURE
# ============================================

print("============================================")
print("CAPTURING VOLATILE EVIDENCE")
print("============================================")

volatile_file = f"{evidence_folder}/volatile_evidence_{timestamp}.txt"

with open(volatile_file, "w", encoding="utf-8") as file:

    file.write("VOLATILE EVIDENCE CAPTURE\n")
    file.write("=" * 50 + "\n")

    # System information
    file.write("\n[System Information]\n")
    file.write(f"Operating System: {platform.system()}\n")
    file.write(f"System Version: {platform.version()}\n")
    file.write(f"Machine: {platform.machine()}\n")

    # Hostname
    file.write("\n[Hostname]\n")
    file.write(f"{socket.gethostname()}\n")

    # Current time
    file.write("\n[Capture Time]\n")
    file.write(f"{datetime.now()}\n")

    # Network information
    file.write("\n[Network Information]\n")

    try:
        hostname = socket.gethostname()
        ip_address = socket.gethostbyname(hostname)

        file.write(f"Hostname: {hostname}\n")
        file.write(f"IP Address: {ip_address}\n")

    except Exception as e:
        file.write(f"Network information unavailable: {e}\n")

    # Running processes
    file.write("\n[Running Processes]\n")

    try:
        process_output = subprocess.check_output(
            ["ps", "aux"],
            text=True
        )

        file.write(process_output)

    except Exception as e:
        file.write(f"Process information unavailable: {e}\n")


print("Volatile evidence captured successfully.")

# ============================================
# 2. NON-VOLATILE EVIDENCE CAPTURE
# ============================================

print("\n============================================")
print("CAPTURING NON-VOLATILE EVIDENCE")
print("============================================")

print("Upload a file to simulate non-volatile evidence.")

uploaded = files.upload()

evidence_file = list(uploaded.keys())[0]

# Copy uploaded evidence to forensic evidence folder
non_volatile_file = f"{evidence_folder}/non_volatile_{evidence_file}"

with open(evidence_file, "rb") as source:
    data = source.read()

with open(non_volatile_file, "wb") as destination:
    destination.write(data)

print("Non-volatile evidence copied successfully.")

# ============================================
# 3. HASH NON-VOLATILE EVIDENCE
# ============================================

def calculate_sha256(filename):

    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:
        for block in iter(lambda: file.read(4096), b""):
            sha256.update(block)

    return sha256.hexdigest()


non_volatile_hash = calculate_sha256(non_volatile_file)

# ============================================
# 4. CREATE EVIDENCE REPORT
# ============================================

report_file = f"{evidence_folder}/evidence_report.txt"

with open(report_file, "w", encoding="utf-8") as report:

    report.write("DIGITAL FORENSIC EVIDENCE REPORT\n")
    report.write("=" * 50 + "\n")

    report.write("\nVOLATILE EVIDENCE\n")
    report.write("-" * 30 + "\n")
    report.write(f"File: {volatile_file}\n")
    report.write("Examples: Running processes, network connections, system information\n")
    report.write("Priority: Capture immediately before shutdown or restart\n")

    report.write("\nNON-VOLATILE EVIDENCE\n")
    report.write("-" * 30 + "\n")
    report.write(f"File: {non_volatile_file}\n")
    report.write(f"SHA-256 Hash: {non_volatile_hash}\n")
    report.write("Examples: Hard disk files, USB files, forensic images\n")
    report.write("Priority: Preserve and hash for integrity verification\n")

    report.write("\nCapture Time: ")
    report.write(str(datetime.now()))

# ============================================
# 5. DISPLAY SUMMARY
# ============================================

print("\n============================================")
print("EVIDENCE CAPTURE SUMMARY")
print("============================================")

print("\nVOLATILE EVIDENCE:")
print(f"Saved File: {volatile_file}")

print("\nNON-VOLATILE EVIDENCE:")
print(f"Saved File: {non_volatile_file}")
print(f"SHA-256: {non_volatile_hash}")

print("\nREPORT:")
print(f"Saved File: {report_file}")

print("\nEvidence capture completed successfully.")
```
