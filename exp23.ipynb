```python
import hashlib
import shutil
from datetime import datetime

print("=" * 65)
print("        END-TO-END DIGITAL FORENSIC WORKFLOW")
print("=" * 65)

# -------------------------------------------------
# Step 1: Identification
# -------------------------------------------------
print("\n[1] Identification")
print("Evidence identified: suspect_file.txt")

# -------------------------------------------------
# Step 2: Preservation
# -------------------------------------------------
with open("suspect_file.txt", "w") as file:
    file.write("This is sample digital evidence for forensic analysis.")

print("[2] Preservation")
print("Original evidence preserved.")

# -------------------------------------------------
# Step 3: Collection (Forensic Copy)
# -------------------------------------------------
shutil.copy("suspect_file.txt", "evidence_copy.img")
print("[3] Collection")
print("Forensic image created: evidence_copy.img")

# -------------------------------------------------
# Step 4: Hash Verification
# -------------------------------------------------
def calculate_hash(filename):
    sha = hashlib.sha256()

    with open(filename, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha.update(data)

    return sha.hexdigest()

original_hash = calculate_hash("suspect_file.txt")
image_hash = calculate_hash("evidence_copy.img")

print("\n[4] Integrity Verification")
print("Original Hash :", original_hash)
print("Image Hash    :", image_hash)

if original_hash == image_hash:
    print("Integrity Check : PASSED")
else:
    print("Integrity Check : FAILED")

# -------------------------------------------------
# Step 5: Examination
# -------------------------------------------------
print("\n[5] Examination")

with open("evidence_copy.img", "r") as file:
    content = file.read()

print("Recovered Content:")
print(content)

# -------------------------------------------------
# Step 6: Analysis
# -------------------------------------------------
print("\n[6] Analysis")

keyword = "forensic"

if keyword.lower() in content.lower():
    print("Keyword Found :", keyword)
else:
    print("Keyword Not Found")

# -------------------------------------------------
# Step 7: Report Generation
# -------------------------------------------------
print("\n[7] Incident Report")
print("-" * 40)
print("Report Time :", datetime.now())
print("Evidence File : suspect_file.txt")
print("Forensic Image : evidence_copy.img")
print("Hash Verified :", original_hash == image_hash)
print("Keyword Found :", keyword)
print("Status : Investigation Completed")

print("\n" + "=" * 65)
print("FORENSIC WORKFLOW COMPLETED SUCCESSFULLY")
print("=" * 65)
```
