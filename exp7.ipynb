```python
def identify_file_type(filename):
    # Common file signatures (magic numbers)
    magic_numbers = {
        b"\xFF\xD8\xFF": "JPEG Image",
        b"\x89PNG\r\n\x1a\n": "PNG Image",
        b"%PDF": "PDF Document",
        b"PK\x03\x04": "ZIP Archive / DOCX / XLSX",
        b"GIF87a": "GIF Image",
        b"GIF89a": "GIF Image",
        b"\x25\x50\x44\x46": "PDF Document",
        b"\x7FELF": "Linux ELF Executable",
        b"MZ": "Windows EXE/DLL",
        b"RIFF": "RIFF Multimedia File"
    }

    with open(filename, "rb") as file:
        # Read first 16 bytes of the file
        file_signature = file.read(16)

    for signature, file_type in magic_numbers.items():
        if file_signature.startswith(signature):
            return file_type

    return "Unknown File Type"


# Get filename from the user
filename = input("Enter the file path: ")

try:
    file_type = identify_file_type(filename)
    print("Detected File Type:", file_type)

except FileNotFoundError:
    print("Error: File not found.")

except PermissionError:
    print("Error: Permission denied.")
```
