```python
import hashlib
import shutil

# Function to calculate SHA-256 hash
def calculate_hash(file_path):
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha256.update(data)

    return sha256.hexdigest()

# -----------------------------
# Create a sample evidence file
# -----------------------------
with open("original_evidence.txt", "wb") as file:
    file.write(b"This is a sample digital evidence for bit stream copy verification.")

print("Original Evidence File Created Successfully.")

# -----------------------------
# Perform Bit Stream Copy
# -----------------------------
shutil.copyfile("original_evidence.txt", "bitstream_copy.img")

print("Bit Stream Copy Completed.")

# -----------------------------
# Calculate Hash Values
# -----------------------------
original_hash = calculate_hash("original_evidence.txt")
copy_hash = calculate_hash("bitstream_copy.img")

print("\nSHA-256 Hash Values")
print("-" * 45)
print("Original File Hash :", original_hash)
print("Copied File Hash   :", copy_hash)

# -----------------------------
# Verify Integrity
# -----------------------------
print("\nVerification Result")
print("-" * 45)

if original_hash == copy_hash:
    print("SUCCESS: Bit Stream Copy Verified.")
    print("Both files are identical.")
else:
    print("FAILED: Copy Verification Failed.")
    print("Files are different.")

print("\nBit Stream Copy Verification Completed.")
```
