```python
import shutil
import hashlib
import os

# Function to calculate SHA-256 hash
def calculate_hash(filename):
    sha256 = hashlib.sha256()

    with open(filename, "rb") as f:
        while True:
            data = f.read(4096)
            if not data:
                break
            sha256.update(data)

    return sha256.hexdigest()

# -----------------------------
# Create a sample evidence file
# -----------------------------
with open("evidence.txt", "w") as f:
    f.write("This is a sample digital evidence file.")

print("Original Evidence File Created.")

# -----------------------------
# Duplication
# -----------------------------
shutil.copy("evidence.txt", "duplicate_evidence.txt")
print("\nDuplication Completed.")

# -----------------------------
# Imaging (Simulation)
# -----------------------------
shutil.copy("evidence.txt", "evidence_image.img")
print("Forensic Imaging Completed.")

# -----------------------------
# Hash Verification
# -----------------------------
original_hash = calculate_hash("evidence.txt")
duplicate_hash = calculate_hash("duplicate_evidence.txt")
image_hash = calculate_hash("evidence_image.img")

print("\nSHA-256 Hash Values")
print("-" * 40)
print("Original File :", original_hash)
print("Duplicate File:", duplicate_hash)
print("Image File    :", image_hash)

# -----------------------------
# Integrity Verification
# -----------------------------
print("\nVerification Results")
print("-" * 40)

if original_hash == duplicate_hash:
    print("Duplicate Verification : SUCCESS")
else:
    print("Duplicate Verification : FAILED")

if original_hash == image_hash:
    print("Image Verification     : SUCCESS")
else:
    print("Image Verification     : FAILED")

print("\nImaging vs Duplication Demonstration Completed.")
```
