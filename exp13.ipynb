```python
# ============================================
# AUTOMATIC DIGITAL FORENSIC REPORT GENERATOR
# Google Colab Ready
# ============================================

from datetime import datetime
from google.colab import files

# Step 1: Collect investigation findings
case_id = input("Enter Case ID: ")
investigator = input("Enter Investigator Name: ")
incident_type = input("Enter Incident Type: ")
evidence_description = input("Enter Evidence Description: ")
findings = input("Enter Investigation Findings: ")
conclusion = input("Enter Conclusion: ")
recommendations = input("Enter Recommendations: ")

# Step 2: Generate current date and time
report_date = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

# Step 3: Create structured forensic report
report = f"""
============================================================
          DIGITAL FORENSIC INVESTIGATION REPORT
============================================================

CASE INFORMATION
------------------------------------------------------------
Case ID              : {case_id}
Investigator         : {investigator}
Report Date          : {report_date}
Incident Type        : {incident_type}

------------------------------------------------------------
1. EVIDENCE DESCRIPTION
------------------------------------------------------------
{evidence_description}

------------------------------------------------------------
2. INVESTIGATION FINDINGS
------------------------------------------------------------
{findings}

------------------------------------------------------------
3. ANALYSIS
------------------------------------------------------------
The collected evidence was examined using standard digital
forensic investigation procedures. The available evidence was
analyzed to identify relevant digital artifacts, suspicious
activities, and possible indicators of compromise.

------------------------------------------------------------
4. CONCLUSION
------------------------------------------------------------
{conclusion}

------------------------------------------------------------
5. RECOMMENDATIONS
------------------------------------------------------------
{recommendations}

------------------------------------------------------------
6. EVIDENCE INTEGRITY
------------------------------------------------------------
Evidence should be preserved using cryptographic hashing
and proper chain-of-custody procedures.

------------------------------------------------------------
END OF REPORT
============================================================
"""

# Step 4: Display report
print(report)

# Step 5: Save report as a text file
output_file = "digital_forensic_report.txt"

with open(output_file, "w", encoding="utf-8") as file:
    file.write(report)

print("\nReport generated successfully!")
print("File:", output_file)

# Step 6: Download the report
files.download(output_file)
```
