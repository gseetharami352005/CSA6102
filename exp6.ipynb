```python
import hashlib
import shutil
import os

# Create a dummy evidence.txt file if it doesn't exist
if not os.path.exists("evidence.txt"):
    with open("evidence.txt", "w") as f:
        f.write("This is some sample evidence data.\n")

def calculate_hash(filename):
    """Calculate SHA-256 hash of a file."""
    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha256.update(data)

    return sha256.hexdigest()


# Input and forensic copy filenames
source_file = "evidence.txt"
copy_file = "forensic_copy.img"

# Step 1: Calculate hash of the original file
original_hash = calculate_hash(source_file)

# Step 2: Create a bit-stream forensic copy
shutil.copyfile(source_file, copy_file)

# Step 3: Calculate hash of the forensic copy
copy_hash = calculate_hash(copy_file)

# Step 4: Verify integrity
print("Original File SHA-256 Hash:", original_hash)
print("Forensic Copy SHA-256 Hash:", copy_hash)

if original_hash == copy_hash:
    print("\nIntegrity Verification: SUCCESS")
    print("The forensic copy is identical to the original file.")
else:
    print("\nIntegrity Verification: FAILED")
    print("The forensic copy has been modified or corrupted.")
```
